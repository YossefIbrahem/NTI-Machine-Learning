# Data Cleaning duplicated

```python
print('Number of duplicate rows : ', df.duplicated().sum())
df.drop_duplicates(inplace=True)
```

# Data Cleaning missing Value

```python
df.isnull().sum().sort_values(ascending=False).head(20)
```

Columns with more than 50% missing values are dropped (PoolQC, MiscFeature, Alley, Fence, MasVnrType) and Id is not useful

```python
df.drop(['PoolQC', 'MiscFeature', 'Alley', 'Fence', 'MasVnrType', 'Id'], axis=1, inplace=True)

# TotalBsmtSF, GrLivArea and Exterior2nd are exact combinations / copies of other columns, so they are dropped
df.drop(['TotalBsmtSF', 'GrLivArea', 'Exterior2nd'], axis=1, inplace=True)
```

## Replace numerical NaN values

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(missing_values=np.nan, strategy='median')
for col in ['LotFrontage', 'MasVnrArea', 'GarageYrBlt']:
    df[[col]] = imputer.fit_transform(df[[col]])
```

## Replace categorical NaN values

```python
imputer = SimpleImputer(missing_values=np.nan, strategy='most_frequent')
for col in df.select_dtypes(include=['object']).columns:
    df[[col]] = imputer.fit_transform(df[[col]])
```

```python
df.isnull().sum().sum()
```

# Data Encoding

## Ordinal Encoding (quality columns)

```python
from sklearn.preprocessing import OrdinalEncoder

quality_cols = ['ExterQual', 'ExterCond', 'BsmtQual', 'BsmtCond', 'HeatingQC', 'KitchenQual', 'FireplaceQu', 'GarageQual', 'GarageCond']
ord_enc = OrdinalEncoder(categories=[['Po', 'Fa', 'TA', 'Gd', 'Ex']] * len(quality_cols))
df[quality_cols] = ord_enc.fit_transform(df[quality_cols])
```

## Label Encoder

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['Neighborhood'] = le.fit_transform(df['Neighborhood'])
df['CentralAir'] = le.fit_transform(df['CentralAir'])
```

## OneHotEncoder

```python
from sklearn.preprocessing import OneHotEncoder

cat_cols = df.select_dtypes(include=['object']).columns
ohe = OneHotEncoder(handle_unknown='infrequent_if_exist', min_frequency=10, drop='first', sparse_output=False).set_output(transform='pandas')
encoded = ohe.fit_transform(df[cat_cols])
df = pd.concat([df.drop(columns=cat_cols), encoded], axis=1)
df.shape
```

# Outliers

```python
plt.figure(figsize=(6, 5))
plt.boxplot(df['SalePrice'])
plt.title('Sale Price Outliers')
plt.show()
```

## IQR

```python
q1 = df['SalePrice'].quantile(0.25)
q3 = df['SalePrice'].quantile(0.75)
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

df = df[(df['SalePrice'] >= lower) & (df['SalePrice'] <= upper)]
df.reset_index(drop=True, inplace=True)
df.shape
```

# Split data

```python
X = df.drop('SalePrice', axis=1)
y = df[['SalePrice']]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=100)
print('X_train = ', X_train.shape, ' y_train = ', y_train.shape)
print('X_test  = ', X_test.shape, ' y_test  = ', y_test.shape)
```

# Scaling

```python
from sklearn.preprocessing import MinMaxScaler

my_scaler = MinMaxScaler(feature_range=(0, 1))
X_train = my_scaler.fit_transform(X_train)
X_test = my_scaler.transform(X_test)
```

# Models

## Linear Regression

```python
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

lr = LinearRegression()
lr.fit(X_train, y_train)

y_pred_lr = lr.predict(X_test)
print('Train R2:', r2_score(y_train, lr.predict(X_train)))
print('Test R2:', r2_score(y_test, y_pred_lr))
print('Test MAE:', mean_absolute_error(y_test, y_pred_lr))
print('Test MSE:', mean_squared_error(y_test, y_pred_lr))
```

## Ridge

```python
ridge = Ridge(alpha=1.0)
ridge.fit(X_train, y_train)

y_pred_ridge = ridge.predict(X_test)
print('Train R2:', r2_score(y_train, ridge.predict(X_train)))
print('Test R2:', r2_score(y_test, y_pred_ridge))
print('Test MAE:', mean_absolute_error(y_test, y_pred_ridge))
print('Test MSE:', mean_squared_error(y_test, y_pred_ridge))
```

## GridSearchCV (Ridge)

```python
from sklearn.model_selection import GridSearchCV

grid = GridSearchCV(estimator=Ridge(), param_grid={'alpha': [0.01, 0.1, 1, 10, 100]}, cv=5, scoring='r2')
grid.fit(X_train, y_train)

best_ridge = grid.best_estimator_
print(grid.best_params_)
y_pred_best = best_ridge.predict(X_test)
print('Test R2:', r2_score(y_test, y_pred_best))
print('Test MAE:', mean_absolute_error(y_test, y_pred_best))
```

## Decision Tree

```python
from sklearn.tree import DecisionTreeRegressor, plot_tree

dt = DecisionTreeRegressor(criterion='squared_error', max_depth=5, min_samples_split=10, min_samples_leaf=5, random_state=100)
dt.fit(X_train, y_train)

y_pred_dt = dt.predict(X_test)
print('Train R2:', r2_score(y_train, dt.predict(X_train)))
print('Test R2:', r2_score(y_test, y_pred_dt))
print('Test MAE:', mean_absolute_error(y_test, y_pred_dt))
print('Test MSE:', mean_squared_error(y_test, y_pred_dt))
```

```python
plt.figure(figsize=(18, 9))
plot_tree(dt, feature_names=X.columns.tolist(), max_depth=3, filled=True, rounded=True, fontsize=8)
plt.title('Decision Tree Regressor (first 3 levels)')
plt.show()
```

## Models comparison

```python
results = pd.Series({
    'Linear Regression': r2_score(y_test, y_pred_lr),
    'Ridge': r2_score(y_test, y_pred_ridge),
    'Ridge (Grid)': r2_score(y_test, y_pred_best),
    'Decision Tree': r2_score(y_test, y_pred_dt),
}).sort_values()
print(results)

plt.figure(figsize=(8, 4))
results.plot(kind='barh', color='teal')
plt.xlabel('Test R2')
plt.title('Models Comparison')
plt.show()
```
