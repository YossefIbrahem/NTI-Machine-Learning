```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv(r'/Users/pawn/Desktop/NTI/Canada.csv')
df.head(10)
```

```python
df.set_index('Country', inplace=True)
```

```python
df.head(10)
```

```python
years = list(map(str, range(1980, 2014)))
print(years)
```

```python
haiti = df.loc['Haiti', years]
haiti.head()
```

```python
haiti = df.loc['Haiti', years]
haiti.plot(kind='line' , color='orange', figsize=(10, 5))
plt.title('Haiti GDP per Capita (1980-2013)', fontsize=16, color='blue')
plt.xlabel('Years')
plt.ylabel('Number of Immigrants')
plt.show()
```

```python
plt.figure(figsize=(20, 10))
plt.plot(haiti.index, haiti.values, color='red', marker='o')
plt.title('Haiti GDP per Capita (1980-2013)', fontsize=16, color='blue')
plt.xlabel('Years')
plt.ylabel('Number of Immigrants')
plt.xticks(rotation=45)
plt.show()
```

```python
sns.set(style='whitegrid')
plt.figure(figsize=(20, 10))
sns.lineplot(data=haiti, marker='o', color='black')
plt.title('Haiti GDP per Capita (1980-2013)', fontsize=16, color='blue')
plt.xlabel('Years')
plt.ylabel('Number of Immigrants')
plt.xticks(rotation=45)
plt.show()
```

```python
df_tot = pd.DataFrame(df[years1].sum(axis=0))

df_tot.reset_index(inplace=True)
df_tot.columns = ['years' , 'Total']
df_tot.head()
```

```python
df_tot.plot(kind='bar' , figsize=(10, 5) )
plt.title('Immigration to canada (1980-2013)', fontsize=16, color='blue')
plt.ylabel('number of Immigrants')
plt.show()
```

```python
df_tot.plot(kind='scatter' , x='years', y='Total', figsize=(20, 10), color='blue')
plt.title('Immigration to canada (1980-2013)', fontsize=16, color='blue')
plt.ylabel('number of Immigrants')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(20,10))
plt.scatter(df_tot['years'], df_tot['Total'], color='blue')
plt.title('Immigration to canada (1980-2013)', fontsize=16, color='blue')
plt.ylabel('number of Immigrants')
plt.show()
```

```python
sns.set(style='whitegrid')
sns.scatterplot(x=df_tot['years'], y=df_tot['Total'], color='blue' )
plt.title('Immigration to canada (1980-2013)', fontsize=16, color='blue')
plt.ylabel('number of Immigrants')
plt.xticks(rotation=45)
plt.show()
```

```python
sns.set(style='white')

plt.figure(figsize=(20,10))
sns.countplot(x='Region' , hue='DevName' , data=df,order=df['Region'].value_counts().index)
plt.title('Immigration to canada (1980-2013)', fontsize=16, color='blue')
plt.ylabel('number of Immigrants')
plt.xticks(rotation=90)
plt.show()
```
