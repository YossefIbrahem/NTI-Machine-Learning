```python
from sklearn.impute import SimpleImputer

impute = SimpleImputer(missing_values=np.nan, strategy='median')
impute.fit(df[['years_of_experience']])
df[['years_of_experience']] = impute.transform(df[['years_of_experience']])
```

```python
impute = SimpleImputer(missing_values=np.nan, strategy='most_frequent')
impute.fit(df[['company_size']])
df[['company_size']] = impute.transform(df[['company_size']])
```

```python
df.isnull().sum().sort_values(ascending=False).head(12)
```

```python
for column in df.columns:
    if df[column].isna().any():
        if column in {'work_year', 'remote_ratio'} or not pd.api.types.is_numeric_dtype(df[column]):
            strategy = 'most_frequent'
        else:
            strategy = 'median'

        imputer = SimpleImputer(strategy=strategy)
        df[[column]] = imputer.fit_transform(df[[column]])

df.isna().sum()
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['experience_level'] = le.fit_transform(df['experience_level'])
df['salary_currency'] = le.fit_transform(df['salary_currency'])
```

```python
le.inverse_transform(df['salary_currency'])
```

```python
from sklearn.preprocessing import OrdinalEncoder
ord_enc = OrdinalEncoder(categories=[['S' , 'M' , 'L']])
df['company_size'] = ord_enc.fit_transform(df[['company_size']])
```

```python
from sklearn.preprocessing import OneHotEncoder
one = OneHotEncoder(
    handle_unknown='ignore',
    sparse_output=False
    
).set_output(transform="pandas")
one_df = one.fit_transform(df[['employment_type']])

df = pd.concat([df , one_df] , axis=1)
```

## Additional visualizations (AI jobs salaries)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
sns.set_theme(style='whitegrid')
ai = pd.read_csv(r'/Users/pawn/Desktop/NTI/salaries (1) copy.csv')
ai.shape
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(ai['salary_in_usd'].dropna(), bins=50, color='steelblue')
plt.title('Distribution of Salary (USD)')
plt.show()
```

```python
order = ['EN', 'MI', 'SE', 'EX']
plt.figure(figsize=(8, 5))
sns.countplot(x='experience_level', data=ai, order=order, color='teal')
plt.title('Employees per Experience Level')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x='experience_level', y='salary_in_usd', data=ai, order=order)
plt.title('Salary by Experience Level')
plt.show()
```

```python
top = ai.groupby('job_title')['salary_in_usd'].mean().sort_values(ascending=False).head(10)
plt.figure(figsize=(9, 5))
sns.barplot(x=top.values, y=top.index, color='slateblue')
plt.title('Top 10 Job Titles by Average Salary')
plt.xlabel('Average salary (USD)')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
ai.groupby('work_year')['salary_in_usd'].mean().plot(marker='o', color='darkorange')
plt.title('Average Salary per Work Year')
plt.ylabel('Average salary (USD)')
plt.show()
```

```python
plt.figure(figsize=(6, 6))
ai['employment_type'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Employment Type')
plt.ylabel('')
plt.show()
```
