## Additional visualizations (San Francisco salaries)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
sns.set_theme(style='whitegrid')
sf = pd.read_csv(r'/Users/pawn/Desktop/NTI/Salaries.csv', skiprows=[1])
sf.columns = sf.columns.str.rstrip('1')
for col in ['BasePay', 'OvertimePay', 'OtherPay', 'Benefits', 'TotalPay', 'TotalPayBenefits', 'Year']:
    sf[col] = pd.to_numeric(sf[col], errors='coerce')
sf.shape
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(sf['TotalPay'].dropna(), bins=50, color='steelblue')
plt.title('Distribution of Total Pay')
plt.xlabel('Total Pay')
plt.show()
```

```python
top_jobs = sf['JobTitle'].str.title().value_counts().head(10)
plt.figure(figsize=(9, 5))
sns.barplot(x=top_jobs.values, y=top_jobs.index, color='teal')
plt.title('Top 10 Job Titles')
plt.xlabel('Employees')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sf.groupby('Year')['TotalPay'].mean().plot(kind='bar', color='slateblue')
plt.title('Average Total Pay per Year')
plt.ylabel('Average Total Pay')
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.boxplot(x='Year', y='TotalPay', data=sf)
plt.title('Total Pay by Year')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sample = sf.dropna(subset=['BasePay', 'TotalPay']).sample(5000, random_state=1)
sns.scatterplot(x='BasePay', y='TotalPay', data=sample, alpha=0.4)
plt.title('Base Pay vs Total Pay')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(sf[['BasePay', 'OvertimePay', 'OtherPay', 'Benefits', 'TotalPay', 'TotalPayBenefits']].corr(), annot=True, fmt='.2f', cmap='YlGnBu')
plt.title('Correlation of Pay Columns')
plt.show()
```
